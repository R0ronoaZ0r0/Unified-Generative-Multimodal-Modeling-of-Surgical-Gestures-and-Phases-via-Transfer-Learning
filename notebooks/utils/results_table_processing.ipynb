```python
import pandas as pd
from pylatex import Document, Section, Command, NoEscape, Package
import os
```

```python
CSV_SAVE_DIR = '../../documents/accuracies/'
os.makedirs(CSV_SAVE_DIR, exist_ok=True)
# CSV_LOAD_PATH = CSV_SAVE_DIR + 'report_results.csv'
CSV_LOAD_PATH = '../../documents/accuracies/report_results.csv'
```

```python
df = pd.read_csv(CSV_LOAD_PATH)
print(df.head())
print(df.columns)
df = df.drop(columns=['lr', 'batch_size', 'hidden_dim'])
print(df.columns)
```

```python
# Split tables based on Task
df_suturing = df[df['task'] == 'Suturing'].drop(columns=['task'])
df_needle = df[df['task'] == 'Needle_Passing'].drop(columns=['task'])
df_knot = df[df['task'] == 'Knot_Tying'].drop(columns=['task'])

# print(len(df_suturing), len(df_needle), len(df_knot))
# print(df_suturing.columns)

# merge kind and type to model_name
df_suturing['model_name'] = df_suturing['kind'] + ' ' + df_suturing['type']
df_needle['model_name'] = df_needle['kind'] + ' ' + df_needle['type']
df_knot['model_name'] = df_knot['kind'] + ' ' + df_knot['type']

# drop kind and type columns
df_suturing = df_suturing.drop(columns=['kind', 'type'])
df_needle = df_needle.drop(columns=['kind', 'type'])
df_knot = df_knot.drop(columns=['kind', 'type'])

# redo column names
df_suturing.columns = df_suturing.columns.str.replace('_', ' ')
df_needle.columns = df_needle.columns.str.replace('_', ' ')
df_knot.columns = df_knot.columns.str.replace('_', ' ')

# Capitalize first letter of each word in column names
df_suturing.columns = df_suturing.columns.str.title()
df_needle.columns = df_needle.columns.str.title()
df_knot.columns = df_knot.columns.str.title()

# reset index
df_suturing = df_suturing.reset_index(drop=True)
df_needle = df_needle.reset_index(drop=True)
df_knot = df_knot.reset_index(drop=True)

# Reorder columns to have Model Name first
df_suturing = df_suturing[['Model Name', 'Accuracy', 'Top5 Accuracy', 'Weighted F1',  'Train Time']]
df_needle = df_needle[['Model Name', 'Accuracy', 'Top5 Accuracy', 'Weighted F1',  'Train Time']]
df_knot = df_knot[['Model Name', 'Accuracy', 'Top5 Accuracy', 'Weighted F1',  'Train Time']]

print(df_knot.head())
```

```python
# split tables for gesture only
df_suturing_gesture = df_suturing[df_suturing['Model Name'].str.contains('Gesture')].reset_index(drop=True)
df_needle_gesture = df_needle[df_needle['Model Name'].str.contains('Gesture')].reset_index(drop=True)
df_knot_gesture = df_knot[df_knot['Model Name'].str.contains('Gesture')].reset_index(drop=True)
df_gesture = pd.concat([df_suturing_gesture, df_needle_gesture, df_knot_gesture], ignore_index=True)
print(df_gesture)
```

```python
# Split Table Phase models
df_suturing_phase = df_suturing[df_suturing['Model Name'].str.contains('Phase')].reset_index(drop=True)
df_needle_phase = df_needle[df_needle['Model Name'].str.contains('Phase')].reset_index(drop=True)
df_knot_phase = df_knot[df_knot['Model Name'].str.contains('Phase')].reset_index(drop=True)

print(df_knot_phase)
```

```python
# split tables for tl only
df_suturing_tl = df_suturing_phase[:4]
df_needle_tl = df_needle_phase[:4]
df_knot_tl = df_knot_phase[:4]
print(df_knot_tl)
```

```python
# split ablation tables
df_suturing_ablation = pd.concat([df_suturing_phase[:1], df_suturing_phase[4:]], ignore_index=True)
df_needle_ablation = pd.concat([df_needle_phase[:1], df_needle_phase[4:]], ignore_index=True)
df_knot_ablation = pd.concat([df_knot_phase[:1], df_knot_phase[4:]], ignore_index=True)
print(df_knot_ablation)
```

```python
# split tables for grs only
df_suturing_grs = df_suturing[df_suturing['Model Name'].str.contains('GRS')].reset_index(drop=True)
df_needle_grs = df_needle[df_needle['Model Name'].str.contains('GRS')].reset_index(drop=True)
df_knot_grs = df_knot[df_knot['Model Name'].str.contains('GRS')].reset_index(drop=True)
df_grs = pd.concat([df_suturing_grs, df_needle_grs, df_knot_grs], ignore_index=True)
df_grs = df_grs.drop(columns=['Top5 Accuracy', 'Weighted F1'])
print(df_grs)
```

```python
# save gesture tables to csv
df_gesture.to_csv(os.path.join(CSV_SAVE_DIR, 'report_gesture_results.csv'), index=False)

# save grs tables to csv
df_grs.to_csv(os.path.join(CSV_SAVE_DIR, 'report_grs_results.csv'), index=False)

# save transfer learning and ablation tables to csv

df_suturing_ablation.to_csv(os.path.join(CSV_SAVE_DIR, 'df_suturing_ablation.csv'), index=False)
df_needle_ablation.to_csv(os.path.join(CSV_SAVE_DIR, 'df_needle_ablation.csv'), index=False)
df_knot_ablation.to_csv(os.path.join(CSV_SAVE_DIR, 'df_knot_ablation.csv'), index=False)

df_suturing_tl.to_csv(os.path.join(CSV_SAVE_DIR, 'df_suturing_tl.csv'), index=False)
df_needle_tl.to_csv(os.path.join(CSV_SAVE_DIR, 'df_needle_tl.csv'), index=False)
df_knot_tl.to_csv(os.path.join(CSV_SAVE_DIR, 'df_knot_tl.csv'), index=False)
```
