```python
import scrapbook as sb
import os
import pandas as pd
```

```python
NOTEBOOKS_DIR = '../../outputs/executed_notebooks/experiments/2025-11-19/20-34/'
```

```python
nbs = [notebook for notebook in os.listdir(NOTEBOOKS_DIR) if notebook.endswith('.ipynb')]
print(len(nbs))
print(nbs[0:5])

nbs_paths = [(nb, os.path.join(NOTEBOOKS_DIR, nb)) for nb in nbs]

print(len(nbs_paths))
print(nbs_paths[0:5])
```

```python
print(nbs[0].split('.')[0].split('_'))
```

```python
results = []

for notebook, notebook_path in nbs_paths:
    nb = sb.read_notebook(notebook_path)
    try:
        result = nb.scraps['results'].data
    except KeyError:
        print(f"No 'results' scrap found in {notebook}")
        continue  # or handle the missing scrap as needed
    if isinstance(result, list):
        results.extend(result)
    else:
        results.append(result)

df = pd.DataFrame(results)
print(df.columns)

# restructure the columns
df = df[['kind', 'task', 'type','lr', 'batch_size', 'hidden_dim', 'accuracy', 'top5_accuracy', 'weighted_f1', 'train_time']]
print(df.columns)
```

```python
df.to_csv('temp.csv', index=False)
```

```python
df1 = pd.read_csv('temp.csv')
df2 = pd.read_csv('../../outputs/reports/2025-10-09/06-23/report_results.csv')

df = pd.concat([df1, df2], axis=0)

df = df[['kind', 'task', 'type','accuracy', 'train_time', 'top5_accuracy', 'weighted_f1', 'lr', 'batch_size', 'hidden_dim']]

df = df.sort_values(by=['task', 'kind', 'type'], ascending=[False, True, False])

df = df.drop_duplicates(subset=['kind', 'task', 'type'], ignore_index=True)
print(df)

save_dir = '../../documents/accuracies/ablation/'
csv_save_path = os.path.join(save_dir, 'report_results_10_09.csv')

df.to_csv(csv_save_path)
```
