```python
import papermill as pm
import os
import subprocess
```

```python
NOTEBOOK_PATHS = ['../preprocessing/preprocessing_jigsaws.ipynb']
TASKS = ['Suturing', 'Needle_Passing', 'Knot_Tying']
OUTPUT_DIR = '../../outputs/executed_notebooks/preprocessing/'

os.makedirs(OUTPUT_DIR, exist_ok=True)
```

```python
def get_output_path(notebook_name, task, output_dir):
    name = notebook_name + '_' + task
    return os.path.join(output_dir, name)



for notebook in NOTEBOOK_PATHS:
    for task in TASKS:
        notebook_name = notebook.split('/')[-1].split('.')[0]
        output_path = get_output_path(notebook_name, task, OUTPUT_DIR)
        output_notebook = output_path + '.ipynb'
        output_pdf = output_path + '.pdf'
        print(output_notebook)
        params = {'TASK': task}
        pm.execute_notebook(
            notebook,
            output_notebook,
            parameters=params,
            progress_bar=True
        )

        try:
            subprocess.run([
                'jupyter', 'nbconvert', '--to', 'pdf', output_notebook, '--output', os.path.abspath(output_pdf)
            ], check=True)
        except subprocess.CalledProcessError as e:
            print(f"Error occurred while converting {output_notebook} to PDF: {e}")
```
